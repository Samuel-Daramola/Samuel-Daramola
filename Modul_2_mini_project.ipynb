```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the sales dataset (replace 'sales_data.csv' with your actual file name)
sales_df = pd.read_csv('/content/Dataset.csv')



#Data Cleaning

# Check for missing values
print("Missing values:\n", sales_df.isnull().sum())

# Handle missing data
sales_df = sales_df.dropna(subset=['Product','Total'])


# Check for duplicates
print("\nNumber of duplicate rows:", sales_df.duplicated().sum())

print("\nMissing values:\n", sales_df.isnull().sum())

# Convert data types
sales_df['Date'] = pd.to_datetime(sales_df['Date'])


# 2. Exploratory Data Analysis (EDA)

# Summary statistics
print("\nSummary Statistics:\n", sales_df.describe())


# Explore relationships between variables
plt.figure(figsize=(8, 6))
sns.barplot(x='Product', y='Price', data=sales_df)
plt.title('Price vs. Product')
plt.show()


# 3. Data Visualization

# Sales trend over time
if 'Date' in sales_df.columns:
  sales_df['Date'] = pd.to_datetime(sales_df['Date'])
  sales_df['Month'] = sales_df['Date'].dt.to_period('M')
  monthly_sales = sales_df.groupby('Month')['Total'].sum()
  plt.figure(figsize=(10, 6))
  monthly_sales.plot(kind='line')
  plt.xlabel('Month')
  plt.ylabel('Total Sales')
  plt.title('Monthly Sales Trend')
  plt.show()

# Distribution of sales
plt.figure(figsize=(8, 6))
sns.histplot(sales_df['Total'], kde=True)
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

From the first figure, laptop is the most expensive item in the store

The second plot shows monthly sales value with August and December having the highest sales.

Lastly, price range of most sold product here is between 100 and 1300
